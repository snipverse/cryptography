```python
def vernam(msg, key):
    result = ""
    for i in range(len(msg)):
        result += chr(ord(msg[i]) ^ ord(key[i]))
    return result

msg = input("Enter message: ")
key = input("Enter key (same length): ")

enc = vernam(msg, key)
print("Encrypted:", enc)

dec = vernam(enc, key)
print("Decrypted:", dec)
```
