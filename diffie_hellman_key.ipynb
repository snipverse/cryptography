```python
p = 23
g = 5

a = int(input("Enter private key of A: "))
b = int(input("Enter private key of B: "))

A = pow(g, a) % p
B = pow(g, b) % p

print("Public value from A:", A)
print("Public value from B:", B)

key_A = pow(B, a) % p
key_B = pow(A, b) % p

print("Secret key for A:", key_A)
print("Secret key for B:", key_B)

if key_A == key_B:
    print("Key shared successfully")
else:
    print("Key sharing failed")
```
