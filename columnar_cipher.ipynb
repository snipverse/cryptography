```python
import math

def encrypt_columnar(text, key):
    text = text.replace(" ", "").upper()
    key = key.upper()

    col = len(key)
    row = math.ceil(len(text) / col)

    matrix = [['' for _ in range(col)] for _ in range(row)]

    k = 0
    for i in range(row):
        for j in range(col):
            if k < len(text):
                matrix[i][j] = text[k]
                k += 1
            else:
                matrix[i][j] = 'X'

    key_order = sorted([(char, i) for i, char in enumerate(key)])

    result = ""
    for char, col_index in key_order:
        for i in range(row):
            result += matrix[i][col_index]

    return result


text = input("Enter plaintext: ")
key = input("Enter key: ")

cipher = encrypt_columnar(text, key)

print("\nEncrypted Text:", cipher)
```
